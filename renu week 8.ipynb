```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_student_performance.csv")
df
```

```python
df.head()
```

```python
parent_education = df.groupby(
    "parental level of education"
)[["math score", "reading score", "writing score"]].mean()

print(parent_education)
```

```python
parent_education.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Scores Based on Parental Level of Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.legend(title="Subjects")
plt.show()
```

```python
lunch_performance = df.groupby(
    "lunch"
)[["math score", "reading score", "writing score"]].mean()

print(lunch_performance)
```

```python
lunch_performance.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Scores Based on Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="test preparation course",
    y="math score",
    data=df
)

plt.title("Math Score: Completed vs Not Completed")
plt.xlabel("Test Preparation Course")
plt.ylabel("Math Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="test preparation course",
    y="reading score",
    data=df
)

plt.title("Reading Score: Completed vs Not Completed")
plt.xlabel("Test Preparation Course")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="test preparation course",
    y="writing score",
    data=df
)

plt.title("Writing Score: Completed vs Not Completed")
plt.xlabel("Test Preparation Course")
plt.ylabel("Writing Score")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x="math score",
    y="reading score",
    data=df
)

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x="math score",
    y="writing score",
    data=df
)

plt.title("Math Score vs Writing Score")
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x="reading score",
    y="writing score",
    data=df
)

plt.title("Reading Score vs Writing Score")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
correlation_data = df[
    ["math score", "reading score", "writing score"]
]

correlation_matrix = correlation_data.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```
